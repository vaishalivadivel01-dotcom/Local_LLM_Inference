```python
import pandas as pd
import requests
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
prompts = pd.read_csv("../dataset/prompts.csv")
responses = pd.read_csv("../dataset/llm_responses.csv")
comparison = pd.read_csv("../dataset/response_comparison.csv")

print("Prompts:")
display(prompts)

print("LLM Responses:")
display(responses.head())

print("Response Comparison:")
display(comparison)
```

```python

# Response Length Comparison Chart

plt.figure(figsize=(10, 5))

plt.bar(
    comparison["id"],
    comparison["word_count"]
)

plt.xlabel("Prompt ID")
plt.ylabel("Word Count")
plt.title("LLM Response Length Comparison")

plt.savefig("../dataset/response_length_chart.png")

plt.show()

print("Chart saved successfully!")
```

```python

average_words = comparison["word_count"].mean()

print("Total Prompts:", len(comparison))
print("Average Word Count:", round(average_words, 2))
print("Maximum Word Count:", comparison["word_count"].max())
print("Minimum Word Count:", comparison["word_count"].min())
```
